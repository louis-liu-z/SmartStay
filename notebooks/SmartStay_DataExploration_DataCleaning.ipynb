# Clean bathrooms_text





```python
import re

# Converts text to a clear numeric value
def extract_bathrooms(text):
    if pd.isna(text):
        return None

    match = re.search(r'(\d+\.?\d*)', text)
    if match:
        return float(match.group(1))
    return None

clean_df['bathrooms'] = clean_df['bathrooms_text'].apply(extract_bathrooms)
```

```python
clean_df['bathrooms'] = clean_df['bathrooms'].fillna(clean_df['bathrooms'].median())
clean_df = clean_df.drop(columns=['bathrooms_text'])
```

*   Convert amenities into vectors



```python
clean_df['amenities_clean'] = clean_df['amenities'].str.replace('[{}"]', '', regex=True).fillna('')

from sklearn.feature_extraction.text import TfidfVectorizer
tfidf = TfidfVectorizer(max_features=100, stop_words='english')
amenities_matrix = tfidf.fit_transform(clean_df['amenities_clean'])
```
